```python
import numpy as np 
import pandas as pd
```

```python
df=pd.read_csv('Loan_approval_data_2025.csv')
df.head()
```

```python
# initial exploration

df.info()
```

```python
# checking for null values

df.isnull().sum()
```

```python
df.head()
```

```python
# Creating some New columns 1. age group

df['age_group']=pd.qcut(df['age'],q=4,labels=['Young','Young Adults','Middle-aged','Older adults'])
```

```python
df[['age','age_group']].head()
```

```python
# 2.Credit score segmentation
l=[]
for i in df['credit_score']:
    if i <580:
        l.append('Low')
    elif i>=580 and i<670:
        l.append('Fair')
    elif i>=670 and i<740:
        l.append('Good')
    elif i>=740 :
        l.append('Excellent')

df['credit_tier']=l

df[['credit_score','credit_tier']].head()

```

```python
df=df[['customer_id', 'age', 'age_group','occupation_status', 'years_employed',
       'annual_income', 'credit_score', 'credit_history_years', 'credit_tier',
       'savings_assets', 'current_debt', 'defaults_on_file',
       'delinquencies_last_2yrs', 'derogatory_marks', 'product_type',
       'loan_intent', 'loan_amount', 'interest_rate', 'debt_to_income_ratio',
       'loan_to_income_ratio', 'payment_to_income_ratio', 'loan_status']]

df.head()
```

```python
# checking for any duplicated rows

df.duplicated().sum()
```

```python
# 3. income bucket column

df['income_bucket']=pd.qcut(df['annual_income'],q=3,labels=['Low income','Middle income','High income'])

df[['annual_income','income_bucket']]
```
